```python
# Gikomba Stall Analysis: Wanjiku's Fresh-Produce Business

# Wanjiku runs a fresh-produce stall at Gikomba Market. This notebook uses **core Python only** (no pandas or NumPy) to analyse her last 14 days of sales and her current stock, and to answer: *How is the business doing? Is it growing? What should she restock, and where should she focus?*
```

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,

               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2



# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {

    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},

    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},

    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},

    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},

    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
```

```python
## Task 1: total, average, highest and lowest daily sales

# Approach: `sum()`, `len()`, `max()` and `min()` give the headline numbers.`list.index()` returns the position of the best and worst day. That position is used to look up the day name in `days`, and integer division by 7 (`position // 7 + 1`) gives the week number, since positions 0-6 are week 1.
```

```python

total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_pos = daily_sales.index(highest_sales)
worst_pos = daily_sales.index(lowest_sales)
best_day, best_week = days[best_pos], best_pos // 7 + 1
worst_day, worst_week = days[worst_pos], worst_pos // 7 + 1

print(f"Total sales (14 days): KES {total_sales:,}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest day: KES {highest_sales:,} {best_day}, week {best_week}")
print(f"Lowest day: KES {lowest_sales:,} on {worst_day} in week {worst_week})")
```

```python
## Task 2: Weekly Totals and Week-on-Week growth percentage
# Approach `week_totals()` slices the list into the first 7 days and the last 7 days, and returns both sums as a tuple. The tuple is unpacked into two variables, and growth is calculated as (week 2 − week 1) ÷ week 1 × 100.
```

```python

def week_totals(sales):
  """Return (week 1 total, week 2 total) for a list of 14 daily sales fogures
  Uses slicing: the first 7 values are week 1 and the next 7 are week 2."""

  return sum(sales[:7]), sum(sales[7:14])

week1_total, week2_total = week_totals(daily_sales)
growth_pct = (week2_total - week1_total) / week1_total * 100

print(f"Week 1 total: KES {week1_total:,}")
print(f"Week 2 total: KES {week2_total:,}")
print(f"Week-on-Week growth: {growth_pct:.2f}%")
```

```python
## Task 3: Classifying Each Day as Strong, Normal or Weak
# Approach: `classify_day()` compares a day's sales with the 14-day average. More than 20% above average is "Strong", more than 20% below is "Weak", and anything else is "Normal". The function is applied to every day, and a dictionary counting pattern (`counts.get(label, 0) + 1`) tallies each category.
```

```python

def classify_day(amount, average):
  """Classify a day's sales against the average. Returns "Strong" if amount is
  more than 20% above average, "Weak" if more than 20% below average, otherwise
  "Normal"."""
  if amount > average * 1.2:
    return "Strong"
  elif amount < average * 0.8:
    return "Weak"
  else:
    return "Normal"

day_labels = [classify_day(amount, average_sales) for amount in daily_sales]
day_counts = {}
for label in day_labels:
  day_counts[label] = day_counts.get(label, 0) + 1
for i, (day, amount, label) in enumerate(zip(days, daily_sales, day_labels), start=1):
  print(f"Day {i:>2} ({day}): KES {amount:>6,} -> {label}")

print("\nCounts:", day_counts)
```

```python
## Task 4: Weekday vs Weekend Average Sales

# Approach: Each sale is paired with its day name using `zip()`. Mon-Fri sales go into a weekday list and Sat-Sun sales into a weekend list. The two averages are compared as a percentage difference relative to the weekday average.
```

```python

weekday_sales = []
weekend_sales = []
for day, amount in zip(days, daily_sales):
  if day in ("Sat", "Sun"):
    weekend_sales.append(amount)
  else:
    weekday_sales.append(amount)

weekday_avg = sum(weekday_sales) / len(weekday_sales)
weekend_avg = sum(weekend_sales) / len(weekend_sales)
weekend_uplift_pct = (weekend_avg - weekday_avg) / weekday_avg * 100

print(f"Weekday average (Mon-Fri): KES {weekday_avg:,.2f}")
print(f"Weekend average (Sat-Sun): KES {weekend_avg:,.2f}")
print(f"Weekend average is {weekend_uplift_pct:.2f}% higher than weekday average")
```

```python
## Task 5: Reorder List and Gross Margins

# Approach: `get_reorder_list()` loops over `stock.items()` and collects every product whose quantity is at or below its reorder level. A second helper calculates gross margin % as (price − cost) ÷ price × 100 for each product.
```

```python

def get_reorder_list(stock_dict):
  """Return a list of product names whose quantity is at or below the reorder level."""
  to_reorder = []
  for product, info in stock_dict.items():
    if info["qty"] <= info["reorder"]:
      to_reorder.append(product)
  return to_reorder

def gross_margin_pct(price, cost):
  """Return gross margin as a percentage: (price - cost) / price * 100."""
  return (price - cost) / price * 100

reorder_list = get_reorder_list(stock)
margins = {product: gross_margin_pct(info["price"], info["cost"])
            for product, info in stock.items()}

print("Products to reorder:", ", ".join(reorder_list))
print()
for product, margin in margins.items():
  print(f"|{product:<10} margin: {margin:.2f}%")
```

```python
## Task 6: Stall Performance Report

# Approach: All results from Tasks 1-5 are brought together in one report built with f-strings. Column widths (`:<`, `:>`) keep everything aligned, and money is shown as `KES` with thousands separators (`:,`).
```

```python

width = 56
line = "=" * width
thin = "-" * width

print(line)
print("STALL PERFORMANCE REPORT".center(width))
print("Wanjiku's Fresh-produce Stall, Gikomba Market".center(width))
print(line)

print("\nSALES SUMMARY (14 days)")
print(thin)
print(f"{'Total sales':<26}{'KES ' + format(total_sales, ','):>30}")
print(f"{'Average daily sales':<26}{'KES ' + format(average_sales, ',.2f'):>30}")
print(f"{'Best day':<26}{f'{best_day}, week {best_week} (KES {highest_sales:,})':>30}")
print(f"{'Worst day':<26}{f'{worst_day}, week {worst_week} (KES {lowest_sales:,})':>30}")

print("\nGROWTH")
print(thin)
print(f"{'Week 1 total':<26}{'KES ' + format(week1_total, ','):>30}")
print(f"{'Week 2 total':<26}{'KES ' + format(week2_total, ','):>30}")
print(f"{'Week-on-Week growth':<26}{format(growth_pct, '.2f') + '%' :>30}")

print("\nDAY CLASSIFICATION")
print(thin)
for label in ("Strong", "Normal", "Weak"):
  print(f"{label + ' days':<26}{day_counts.get(label, 0):>30}")

print("\nWEEKDAY vs WEEKEND (average daily sales)")
print(thin)
print(f"{'Weekday (Mon-Fri)':<26}{'KES ' + format(weekday_avg, ',.2f'):>30}")
print(f"{'Weekend (Sat-Sun)':<26}{'KES ' + format(weekend_avg, ',.2f'):>30}")
print(f"{'Weekend uplift':<26}{format(weekend_uplift_pct, '.2f') + '%':>30}")

print("\nSTOCK & MARGINS")
print(thin)
print(f"{'Product':<12}{'Qty (kg)':>10}{'Reorder':>10}{'Margin':>10}{'Status':>14}")
for product, info in stock.items():
  status = 'REORDER' if product in reorder_list else "OK"
  print(f"{product:<12}{info['qty']:>10}{info['reorder']:>10}"
        f"{margins[product]:>9.2f}%{status:>14}")

print("\nRestock now:", ", ".join(reorder_list))
print(line)
```
